```python
monthly_consumption = [320, 300, 280, 350, 400, 420]
price_per_kwh = 0.15

def total_consumption(consumption_list):
    total = 0
    for value in consumption_list:
        total += value
    return total

def average_consumption(consumption_list):
    return total_consumption(consumption_list) / len(consumption_list)

def total_cost(consumption_list, price):
    return total_consumption(consumption_list) * price

print("Electricity Consumption Report")
print("------------------------------")
print(f"Total consumption: {total_consumption(monthly_consumption)} kWh")
print(f"Average consumption: {average_consumption(monthly_consumption)} kWh")
print(f"Total cost: {total_cost(monthly_consumption, price_per_kwh)} EUR")
```
